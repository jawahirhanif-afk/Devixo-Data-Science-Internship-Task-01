```python
import pandas as pd

df = pd.read_csv('supermarket_sales - Sheet1.csv')

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python
# Convert Date into datetime format
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y')

# Convert Time into time format
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M').dt.time

# Identify numerical and categorical columns
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns
categorical_columns = df.select_dtypes(include=['object']).columns

print("Date and Time converted successfully!")

print("\nNumerical Columns:")
print(list(numerical_columns))

print("\nCategorical Columns:")
print(list(categorical_columns))
```

```python
# Calculate descriptive statistics
numerical_summary = df[numerical_columns].agg(
    ['mean', 'median', 'min', 'max', 'std']
)

# Display results
display(numerical_summary)
```

```python
categorical_columns = df.select_dtypes(
    include=['object']
).columns

print("Categorical Features:")
print(list(categorical_columns))
```

```python
for col in categorical_columns:
    print("\nDistribution of", col)
    print(df[col].value_counts())
```

```python
for col in categorical_columns:
    print("\nPercentage Distribution of", col)
    percentages = df[col].value_counts(normalize=True) * 100
    print(percentages.round(2))
```

```python
import matplotlib.pyplot as plt

# Repeat for every numerical feature
for col in numerical_columns:

    # Create a new figure
    plt.figure(figsize=(7, 4))

    # Draw the boxplot
    plt.boxplot(df[col])

    # Add title and axis label
    plt.title(f'Boxplot of {col}')
    plt.ylabel(col)

    # Adjust spacing
    plt.tight_layout()

    # Display graph
    plt.show()
```

```python
import matplotlib.pyplot as plt

for col in numerical_columns:
    plt.figure(figsize=(7, 4))
    plt.hist(df[col], bins=20, edgecolor='black')

    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    plt.tight_layout()
    plt.show()
```

```python
# Calculate Q1 and Q3
Q1 = df[numerical_columns].quantile(0.25)
Q3 = df[numerical_columns].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify and count outliers
outlier_counts = (
    (df[numerical_columns] < lower_bound) |
    (df[numerical_columns] > upper_bound)
).sum()

# Display results
print("Outliers in each numerical feature:")
display(outlier_counts)
```

```python
# Identify rows with outliers in Total
outlier_rows = df[
    (df['Total'] < lower_bound['Total']) |
    (df['Total'] > upper_bound['Total'])
]

# Display important details
display(
    outlier_rows[
        ['Invoice ID', 'Branch', 'Product line',
         'Quantity', 'cogs', 'Tax 5%', 'Total']
    ]
)
```

The IQR method identified 9 potential outliers in Tax 5%, Total, cogs, and gross income. Upon investigation, all 9 transactions had a quantity of 10 and relatively high transaction values. These records appear to represent genuine high-value purchases rather than data errors. Therefore, no outliers were removed from the dataset.

```python
import seaborn as sns
import matplotlib.pyplot as plt

correlation_data = df[
    ['Unit price', 'Quantity', 'Tax 5%', 'Total',
     'cogs', 'gross income', 'Rating']
]

correlation_matrix = correlation_data.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.show()
```

### Correlation Analysis

The correlation matrix was used to understand the relationships between numerical features in the supermarket sales dataset.

**Key Findings:**

1. Quantity has a positive correlation of 0.71 with Total, indicating that transactions with more items generally have higher total sales.
2. Unit price has a positive correlation of 0.63 with Total, showing that higher unit prices are associated with higher transaction values.
3. Quantity has a positive correlation of 0.71 with Tax 5% and gross income.
4. Rating has very weak correlations with the other numerical features, with values ranging from -0.01 to -0.04.
5. There are no strong negative correlations between the selected numerical features.

**Important Observation:**

Tax 5% and gross income have identical values in this dataset. Total is calculated using cost of goods sold (cogs) plus the tax amount. Therefore, some very high correlations are caused by the dataset's mathematical calculations.

**Conclusion:**

Quantity, unit price, and financial features show positive relationships with transaction value. Customer ratings have almost no linear relationship with the numerical sales features in this dataset.

## Part 7: Business Recommendations

Based on the analysis and business insights, the following recommendations are suggested:

1. **Improve the performance of Health and Beauty:** This product line generated the lowest total sales. Management should investigate customer preferences, product availability, pricing, and customer feedback to identify opportunities for improvement.

2. **Study Branch C's performance:** Branch C generated the highest total sales and average transaction value. Management should examine its successful practices and consider applying suitable strategies to other branches.

3. **Strengthen the membership program:** Members generated slightly higher total sales than normal customers. The supermarket could introduce member-exclusive offers and loyalty rewards to encourage repeat purchases.

4. **Maintain convenient payment options:** Cash generated the highest total sales, while E-wallet had the highest number of transactions. The supermarket should continue supporting both methods and monitor payment preferences.

5. **Improve the Home and Lifestyle customer experience:** This product line had the highest average transaction value but the lowest average customer rating. Management should review customer feedback and investigate possible improvements in product quality and service.

### Recommendation Summary

These recommendations aim to improve product-line performance, learn from successful branches, strengthen customer loyalty, support convenient payments, and improve customer satisfaction.
